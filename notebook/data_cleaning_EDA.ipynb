```python
df.drop_duplicates(inplace=True)
print("Shape After Removing Duplicates:", df.shape)
```

```python
# remove rows without valid coordinates
df = df.dropna(subset=["latitude", "longitude"])

# Keep only coordinates within Chicago bounding box (to remove data errors)
df = df[(df["latitude"].between(41.5, 42.3)) & (df["longitude"].between(-88.0, -87.0))]

print("Shape After Filtering Out-of-Bounds Coordinates:", df.shape)
```

```python
df["date"] = pd.to_datetime(df["date"], errors="coerce")
```

```python
# clean spaces, convert to lowercase, replace spaces with underscore
cat_cols = df.select_dtypes(include=['object']).columns.tolist()
for c in cat_cols:
    df[c] = df[c].str.strip().str.lower().replace(" ","_")
```

```python
# fill the missing values - numeric → median; categorical → 'missing'
num_cols = df.select_dtypes(include=np.number).columns.tolist()

for c in num_cols:
    df[c] = df[c].fillna(df[c].median())

for c in cat_cols:
    df[c] = df[c].fillna('missing')
```

```python
violent_crimes = [
    'homicide', 'robbery', 'assault', 'battery', 'criminal sexual assault'
]

df["crime_severity"] = df["primary_type"].apply(
    lambda x: 3 if x in violent_crimes else (2 if "theft" in x or "burglary" in x else 1)
)
```

```python
# handle these column as binary varibles
df["arrest"] = df["arrest"].astype(int)
df["domestic"] = df["domestic"].astype(int)
```

```python
path = r"D:\Percia_MTech\GUVI\python\Projects\patrol_IQ\data\chicago_crime_cleaned.csv"
df = pd.read_csv(path,low_memory=False)
df.head()
```

```python
# crime count by year - trend over time using line plot
plt.figure(figsize=(12,6))
df['year'].value_counts().sort_index().plot(kind='line')
plt.title("Crime Count by Year")
plt.xlabel("Year")
plt.ylabel("Number of Crimes")
plt.show()
```

```python
# crime trend by month
df['date'] = pd.to_datetime(df['date'])
df['month'] = df['date'].dt.month  
plt.figure(figsize=(12,6))
df['month'].value_counts().sort_index().plot(kind='bar')
plt.title("Crime Count by Month")
plt.xlabel("Month")
plt.ylabel("Number of Crimes")
plt.show()
```

```python
# top 10 crime types
plt.figure(figsize=(12,6))
df['primary_type'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Crime Types")
plt.xlabel("Crime Type")
plt.ylabel("Occurrences")
plt.show()
```

```python
# district level crime distribution
plt.figure(figsize=(12,6))
df['district'].value_counts().sort_index().plot(kind='line')
plt.title("Crimes by Police District")
plt.xlabel("District")
plt.ylabel("Count")
plt.show()
```

```python
# domestic vs non domestic crime ratio
plt.figure(figsize=(6,6))
df['domestic'].value_counts().plot(kind='pie', labels=['Non-Domestic', 'Domestic'], autopct='%1.1f%%')
plt.title("Domestic Involvement in Crimes")
plt.show()
```

```python
arrest_rate = df['arrest'].mean() * 100
print(f"Average Arrest Rate: {arrest_rate:.2f}%")
```

```python
# correlation heatmap of numerical features
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=False, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
