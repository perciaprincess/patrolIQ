```python
import pandas as pd
import random
from urllib.parse import quote

samples = []
for i in range(12):  # 12 batches → 12 * 50,000 = 600,000
    offset = random.randint(0, 8400000 - 50000)
    query = (
        "SELECT id, case_number, date, block, iucr, primary_type, description, location_description, "
        "arrest, domestic, beat, district, ward, community_area, fbi_code, x_coordinate, y_coordinate, "
        "year, updated_on, latitude, longitude, location "
        f"LIMIT 50000 OFFSET {offset}"
    )
    encoded_query = quote(query)
    url = f"https://data.cityofchicago.org/resource/ijzp-q8t2.csv?$query={encoded_query}"
    print("Fetching batch with offset:", offset)
    samples.append(pd.read_csv(url))

df = pd.concat(samples, ignore_index=True)
df.to_csv("crime_600k_multi_offset_sample.csv", index=False)
print("✅ Stratified sample saved successfully!")
```

```python
import pandas as pd
import random
from urllib.parse import quote

BATCH_SIZE = 50000
NUM_BATCHES = 16   # 16 * 50,000 ≈ 800,000 rows
MAX_ROWS = 8400000

data_frames = []

for i in range(NUM_BATCHES):
    offset = random.randint(0, MAX_ROWS - BATCH_SIZE)

    query = (
        "SELECT id, case_number, date, block, iucr, primary_type, description, location_description, "
        "arrest, domestic, beat, district, ward, community_area, fbi_code, x_coordinate, y_coordinate, "
        "year, updated_on, latitude, longitude, location "
        f"LIMIT {BATCH_SIZE} OFFSET {offset}"
    )

    encoded_query = quote(query)
    url = f"https://data.cityofchicago.org/resource/ijzp-q8t2.csv?$query={encoded_query}"

    print(f"📥 Downloading batch {i+1}/{NUM_BATCHES} (offset={offset})...")
    df_batch = pd.read_csv(url)
    data_frames.append(df_batch)

df = pd.concat(data_frames, ignore_index=True)

df.to_csv("chicago_crime_8_lakh_sample.csv", index=False)
print("✅ Saved: chicago_crime_8_lakh_sample.csv")
print("📊 Total Rows:", len(df))
print(df['year'].value_counts().sort_index())
```
